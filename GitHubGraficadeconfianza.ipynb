```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar el archivo
df = pd.read_excel("C:/Users/Marben/Desktop/Modelo MovileNet2 con RB/Inferencias/IMGENES_VALIDACION/reporte de clases.xlsx")

# Crear la gráfica
plt.figure(figsize=(18,6))

# Graficar confianza derecho e izquierdo
plt.plot(df['Sujeto'], df['Confianza derecho'], marker='o', label='Derecho', color='blue')
plt.plot(df['Sujeto'], df['Confianza izquierdo'], marker='o', label='Izquierdo', color='orange')

# Configuración
plt.title('Confianza de clasificación por sujeto')
plt.xlabel('Sujeto')
plt.ylabel('Confianza (%)')
plt.xticks(rotation=90)
plt.legend()
plt.grid(True)
plt.tight_layout()

# Mostrar
plt.show()

```
