```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar la imagen
image = cv2.imread('C:/Users/Marben/Desktop/Python/VA/plantar.jpg')

# Verificar si la imagen se ha cargado correctamente
if image is None:
    print("Error al cargar la imagen.")
else:
    # Convertir la imagen a RGB (para que matplotlib lo maneje correctamente)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Convertir la imagen a escala de grises para crear una máscara
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Crear una máscara binaria donde el fondo sea blanco (valor alto) y el objeto sea negro
    _, mask = cv2.threshold(gray, 240, 255, cv2.THRESH_BINARY_INV)

    # Aplicar la máscara a la imagen original para obtener solo el objeto
    object_only = cv2.bitwise_and(image, image, mask=mask)

    # Convertir la imagen segmentada (solo el objeto) de BGR a CIE L*a*b*
    lab_object = cv2.cvtColor(object_only, cv2.COLOR_BGR2Lab)

    # Combinar la imagen LAB con el fondo original
    background = cv2.bitwise_and(image, image, mask=cv2.bitwise_not(mask))
    result_lab = cv2.add(lab_object, background)

    # Ahora, superponer la imagen original sobre el resultado, conservando el fondo de la imagen combinada
    # Invertimos la máscara para aislar el fondo en la imagen combinada
    mask_inv = cv2.bitwise_not(mask)
    
    # Usamos la imagen original pero solo en las zonas del objeto
    object_from_original = cv2.bitwise_and(image, image, mask=mask)

    # Combinamos el objeto original con el fondo modificado (LAB)
    combined_result = cv2.add(object_from_original, background)

    # Mostrar la imagen original, la combinada y el resultado final
    plt.figure(figsize=(10,5))

    plt.subplot(1, 3, 1)
    plt.imshow(image_rgb)
    plt.title('Imagen Original')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(result_lab, cv2.COLOR_BGR2RGB))
    plt.title('Fondo LAB y objeto en BGR')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(combined_result, cv2.COLOR_BGR2RGB))
    plt.title('Imagen Final Superpuesta')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
